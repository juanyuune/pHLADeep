```python
import pandas as pd
from sklearn.model_selection import train_test_split

df = pd.read_csv("/srv/jupyterlab/workspace/HLA1/fasta/iedb_sequences/all_alleles_9mer.csv")

train_df, test_df = train_test_split(
    df,
    test_size=0.2,
    random_state=42,
    stratify=df["binary_label"]
)

train_df = train_df.reset_index(drop=True)
test_df  = test_df.reset_index(drop=True)

train_df.to_csv("/srv/jupyterlab/workspace/HLA1/fasta/iedb_sequences/train.csv", index=False)
test_df.to_csv("/srv/jupyterlab/workspace/HLA1/fasta/iedb_sequences/test.csv",  index=False)

print(f"Train : {len(train_df):,}  (pos: {train_df['binary_label'].sum():,} / neg: {(train_df['binary_label']==0).sum():,})")
print(f"Test  : {len(test_df):,}   (pos: {test_df['binary_label'].sum():,} / neg: {(test_df['binary_label']==0).sum():,})")
```
